```python
import os
import requests
import pandas as pd
from PIL import Image
from io import BytesIO
from tqdm.notebook import tqdm
```

```python
download_link = "https://data.scryfall.io/unique-artwork/unique-artwork-20250323090433.json"
```

```python
data = pd.read_json(download_link, dtype_backend="pyarrow")
data
```

```python
data = data[["name", "id", "image_uris"]]
data.loc[:, "image_uri"] = data.pop("image_uris").apply(lambda x: None if x != x else x["art_crop"])
```

```python
def get_image_array(url: str):

    response = requests.get(url)
    img = Image.open(BytesIO(response.content))
    return img

output_dir = os.path.join("..", "data", "artwork")
os.makedirs(output_dir, exist_ok=True)
```

```python
import os
from PIL.Image import BILINEAR


for _, row in tqdm(data.iterrows(), total=len(data)):
    id = row["id"]
    if row["image_uri"] is not None:
        image = get_image_array(row["image_uri"])
        image.save(os.path.join(output_dir, f"{id}.jpg"))
```
